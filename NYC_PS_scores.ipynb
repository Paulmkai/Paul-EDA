```python
#Task
"""
Analyze NYC school SAT data to identify schools with top math scores (≥80% of 800),
rank the top 10 schools by total SAT score, and determine the borough 
with the highest variability in total SAT performance.
"""
```

```python
import pandas as pd
```

```python
schools = pd.read_csv('schools.csv')
print(schools.head())
```

```python
# Threshold for best math scores
math_threshold = 0.8 * 800

best_math_scores = (schools[schools["average_math"] > math_threshold][["school_name", "average_math"]].sort_values("average_math", ascending=False))
print(best_math_scores)
```

```python
# Create total SAT score
schools["total_SAT"] = (schools["average_math"] + schools["average_reading"] + schools["average_writing"])
print(schools["total_SAT"].head())
```

```python
#list 10 top schools
top_10_schools = (schools[["school_name", "total_SAT"]].sort_values("total_SAT", ascending=False).head(10))
top_10_schools
```

```python
# Group by borough and calculate required statistics
borough_stats = (
    schools.groupby("borough")
    .agg(
        num_schools = ("school_name", "count"),
         average_SAT = ("total_SAT", "mean"),
         std_SAT = ("total_SAT", "std")
    ).round(2).reset_index())
```

```python
print(borough_stats)
```

```python
# Find borough with the largest standard deviation
largest_std_dev = borough_stats[borough_stats["std_SAT"] == borough_stats["std_SAT"].max()]
print(largest_std_dev)
```
